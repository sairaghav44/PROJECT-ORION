```python
#exercise 1

class student:
    
    def __init__(self,name, age):
        self.name= name
        self.age = age

    def introduce(self):
        print(f" his name is {self.name} ,he's {self.age} years old!")

    def update_age(self, new_age):
        self.age = new_age
        
stud1 = student("niggesh",18)
stud2 = student("raghav",28)
 
stud1.introduce()
stud2.introduce()
stud1.update_age(24)
stud1.introduce()
stud2.introduce()



     
```

```python
class BankAccount:

    def __init__(self, name, balance=1000):
        self.name = name
        self.balance = balance
    def deposit(self, amount):
        if amount >0:
            self.balance  += amount
            print("Deposit succesful!")
            print(f"Your balance is {self.balance}")
        else :
            print("enter valid amount!!!")

    def withdraw(self, amount):
        if amount > 0:
            if amount <= self.balance:
                self.balance -= amount
                print("Withdrawal successful!")
                print(f"Your balance is {self.balance}")
            else:
                print("Insufficient balance!")
        else:
            print("Enter a valid amount!")
    def display_balance(self):
        print(f"username : {self.name}")
        print(f"balance: {self.balance}")
person1 = BankAccount("raghav" , 20000)
person1.deposit(10000)
print("                       ")
print("-----user details-----")
person1.display_balance()
        
```
